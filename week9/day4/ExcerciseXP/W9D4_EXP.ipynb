```python
# TODO: fill the comparison table
comparison_table = """| Openness level | What is open | What is closed or unclear | Impact on retraining/modifying |
|---|---|---|---|
| Fully Open | Architecture, source code, weights, training method, and often data documentation | Some data details may still be incomplete or legally restricted | Best option for deep inspection, modification, retraining, and governance |
| Weights Released | Model weights and basic usage documentation | Full training data, training code, and training decisions may be unavailable | Good for local deployment and fine-tuning, but limited for full auditing |
| Architecture Only | Model design, paper, and high-level method | Trained weights, training data, and often training code | Useful for research and reimplementation, but expensive and difficult to use directly |
"""

print(comparison_table)
```

```python
# TODO: your measured system specs
import platform
import shutil

try:
    import psutil
    ram_gb = round(psutil.virtual_memory().total / (1024 ** 3), 1)
except ImportError:
    ram_gb = "Unknown - install psutil or check manually"

free_disk_gb = round(shutil.disk_usage(".").free / (1024 ** 3), 1)

system_specs = {
    "ram_gb": ram_gb,
    "free_disk_gb": free_disk_gb,
    "os": f"{platform.system()} {platform.release()}",
}

system_specs
```

```python
# TODO: readiness table
ram_value = system_specs["ram_gb"]

if isinstance(ram_value, (int, float)):
    ram_status = "Yes" if ram_value >= 16 else "No"
else:
    ram_status = "Unknown"

disk_status = "Yes" if system_specs["free_disk_gb"] >= 20 else "No"
os_status = "Yes" if platform.system() in ["Windows", "Linux", "Darwin"] else "Unknown"

readiness_table = f"""
| Requirement | Your System Specs | Meets Requirement? |
|---|---|---|
| RAM >= 16 GB | {system_specs["ram_gb"]} GB | {ram_status} |
| Free Disk Space >= 20 GB | {system_specs["free_disk_gb"]} GB | {disk_status} |
| OS: Linux / Windows / macOS / WSL2 | {system_specs["os"]} | {os_status} |
| Model format | GGUF required for llama.cpp | Yes, if using GGUF model |
| Local CPU inference | Possible with quantized models | Depends on RAM and CPU speed |
"""

print(readiness_table)
```
