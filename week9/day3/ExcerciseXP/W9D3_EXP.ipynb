## Exercise 2: Multi-Part Prompt for Quiz Generation
**Scenario**: Create an educational prompt for 7th graders about volcanic eruptions.

### Tasks
- Ask for: 2-bullet summary + 3 MCQs (1 correct, 2 distractors).
- Require friendly tone, simple vocab (11?13 y/o), and Google Slides-ready output.
- Explain 3 ways your prompt improves on: `Make a quiz for kids about this article.`

```python
# TODO: write the full multi-part prompt

quiz_prompt = """
You are an educational content creator for 7th grade students.

Create a short educational quiz about volcanic eruptions.

Target audience:
7th grade students, around 11–13 years old.

Tone and language:
- Use a friendly and encouraging tone.
- Use simple vocabulary that 11–13 year old students can understand.
- Avoid complicated scientific terms unless you explain them briefly.

Output format:
Make the answer Google Slides ready.

Create the content in this structure:

Slide 1: Title
- Create a clear and interesting title about volcanic eruptions.

Slide 2: Short Summary
- Write a 2-bullet summary explaining what volcanic eruptions are.
- Keep each bullet short and easy to understand.

Slide 3: Quiz Question 1
- Write one multiple-choice question.
- Include 3 answer choices: A, B and C.
- Only one answer should be correct.
- The other two answers should be believable but incorrect.
- Mark the correct answer clearly.

Slide 4: Quiz Question 2
- Write one multiple-choice question.
- Include 3 answer choices: A, B and C.
- Only one answer should be correct.
- The other two answers should be believable but incorrect.
- Mark the correct answer clearly.

Slide 5: Quiz Question 3
- Write one multiple-choice question.
- Include 3 answer choices: A, B and C.
- Only one answer should be correct.
- The other two answers should be believable but incorrect.
- Mark the correct answer clearly.

Final requirement:
Keep the content clear, short and ready to copy into Google Slides.
"""

quiz_prompt
```

```python
# TODO: explain three improvements over the vague prompt

improvements = [
    "The prompt gives a clear role and target audience, so the answer will be suitable for 7th grade students.",
    "The prompt specifies the exact output: a 2-bullet summary and 3 multiple-choice questions with one correct answer and two distractors.",
    "The prompt adds tone, vocabulary and format requirements, making the result friendly, simple and ready to use in Google Slides."
]

improvements
```

## Exercise 4: Match Prompt to Purpose
Pick a style (Exploratory, Structured, Conversational, Functional) to build a customer-support chatbot prompt.

### Tasks
- Write the full prompt in the chosen style.
- Identify 2 features that prove it matches the style.
- Justify why this style fits better than the others.

```python
# TODO: choose a style and write the prompt

chosen_style = "Functional"

prompt_template = """
You are a customer-support chatbot for an online productivity software company.

Your goal is to help customers solve their problems quickly and clearly.

Instructions:
1. Greet the customer politely.
2. Ask for the key information needed to understand the issue, such as:
   - what problem they are experiencing?
   - what device or browser they are using?
   - whether they see any error message?
   - what steps they already tried?
3. Give clear step-by-step instructions to solve the problem.
4. Use simple and friendly language.
5. If the problem cannot be solved with basic troubleshooting, explain that the issue should be escalated to a human support agent.
6. Do not invent technical information.
7. If information is missing, ask a short follow-up question.

Output format:
- Start with a short friendly response.
- Then provide numbered troubleshooting steps.
- End with either a confirmation question or an escalation message.

Example closing:
“Did these steps solve the issue, or would you like me to connect you with a support agent?”
"""

key_style_features = [
    "The prompt is task-oriented and focuses on solving the customer's problem.",
    "It gives clear rules for collecting information, troubleshooting and escalating the issue when needed."
]

justification = "The functional style fits customer support better than the other styles because the main goal is not to explore ideas or have a casual conversation, but to help the customer complete a practical task and solve a specific problem."

chosen_style, prompt_template, key_style_features, justification
```

## Exercise 6: Hallucination Spotting and Mitigation
**Scenario**: Model added an unsupported claim in a climate/marine article summary.

### Tasks
1) Prompt the model to avoid hallucinations using only provided text.
2) Rewrite with verification language.
3) List 2 mitigation strategies.
4) Name 2 domains where hallucinations are risky and why.

```python
# TODO: baseline anti-hallucination prompt

no_hallucination_prompt = """
You are a careful scientific summarizer.

Summarize the climate/marine article using only the information provided in the text.

Rules:
- Do not add facts, numbers, causes, effects, examples, or conclusions that are not directly supported by the article.
- Do not use outside knowledge.
- If the article does not mention something, do not include it.
- If information is unclear, write: “The article does not provide enough information about this.”
- Keep the summary factual, neutral and concise.

Article text:
...
"""

no_hallucination_prompt
```

```python
# TODO: verification-focused rewrite

verification_prompt = """
You are a fact-checking assistant.

Your task is to summarize the climate/marine article and verify every important claim.

Step 1:
Write a short summary using only the provided article text.

Step 2:
Create a verification table with two columns:
- Claim from the summary
- Evidence from the article

Step 3:
Before finalizing, remove or rewrite any claim that does not have clear support in the article.

Rules:
- Do not add outside information.
- Do not guess.
- Do not include unsupported scientific claims.
- If a claim cannot be verified from the article, write: “Not supported by the provided text.”

Article text:
...
"""

verification_prompt
```

```python
# TODO: list mitigation strategies and risky domains

mitigation_strategies = [
    "Use only the provided source text and clearly forbid outside information.",
    "Ask the model to show evidence for each important claim before finalizing the answer.",
    "Tell the model to say when information is missing instead of guessing or filling gaps."
]

risky_domains = {
    'domain_1': {
        'name': 'Medicine',
        'why': 'Hallucinations can lead to unsafe health advice, incorrect diagnoses or harmful treatment decisions.'
    },
    'domain_2': {
        'name': 'Law and finance',
        'why': 'Hallucinations can cause serious legal, tax, or financial mistakes if people rely on incorrect rules or unsupported claims.'
    },
}

mitigation_strategies, risky_domains
```
