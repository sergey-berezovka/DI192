**Daily Challenge: Precision Prompting for Control & Output Quality**

**Scenario**
You work on the Learning & Communications Team at a global tech company. You’ve been asked to use generative AI to create employee-facing microlearning content for an internal newsletter.

Your task is to turn a paragraph from a recent internal policy update into a concise and clear knowledge snippet. The original policy text is overly formal, dense and full of jargon.

You are an internal communications specialist creating microlearning content for employees.

Rewrite the policy excerpt below into a short, friendly, and clear newsletter snippet.

Requirements:
- Use bullet points.
- Paraphrase the text; do not quote it directly.
- Keep the total response under 75 words.
- Do not add any information that is not in the original text.
- Make the message easy for non-technical employees to understand.

Policy excerpt:
Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications.

```python
# TODO: Step 1: Craft a Prompt That Controls Output Tone, Format and Length

"""
You are an internal communications specialist creating microlearning content for employees.

Rewrite the policy excerpt below into a short, friendly and clear newsletter snippet.

Requirements:
- Use bullet points.
- Paraphrase the text, do not quote it directly.
- Keep the total response under 75 words.
- Do not add any information that is not in the original text.
- Make the message easy for non-technical employees to understand.

Policy excerpt:
Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications.
"""
```

```python
# TODO: Step 2: Evaluate the Output

"""
| Criteria             | Evaluation                                                                                                                                                                                                                       |
| -------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Relevance**        | The output stays focused on the original message: employees must use the approved secure VPN and avoid unsafe access methods.                                                                                                    |
| **Clarity**          | The message is much easier to understand than the original policy text. It uses simple words like “always”, “do not” and “keep safe.”                                                                                            |
| **Structure**        | The information is clearly organized into bullet points, making it easy for employees to scan quickly.                                                                                                                           |
| **Tone**             | The tone is friendly, clear, and appropriate for an internal employee newsletter. It sounds professional but not overly formal.                                                                                                  |
| **Length**           | The output is under 75 words. It has about 40 words, so it meets the length requirement.                                                                                                                                         |
| **Factual Accuracy** | The output does not invent new information. It preserves the key security details: use the approved VPN, avoid unsecured access, do not use unprotected personal devices and protect company data.                               |
"""
```

```python
# TODO: Step 3: Detect and Mitigate Hallucinations

"""
Revised prompt to reduce hallucination risk:

You are an internal communications specialist creating microlearning content for employees.

Rewrite the policy excerpt below into a short, friendly and clear newsletter snippet.

Requirements:
- Use bullet points.
- Paraphrase the text, do not quote it directly.
- Keep the total response under 75 words.
- Only use information from the input text.
- Do not add new recommendations, technologies, examples, tools or policy terms.
- Preserve all critical security details:
  1. Remote access must use the approved secure VPN.
  2. Unsecured connections must not be used.
  3. Personal devices without endpoint protection must not be used.
  4. Proprietary data and sensitive communications must be protected.
- Make the message easy for non-technical employees to understand.

Policy excerpt:
Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data.
"""
```

**Why this prompt is safer:**
It clearly tells the model to paraphrase only the input and not add extra technologies, examples or security advice. It also lists the critical details that must be preserved.

**In what kind of internal communication would quoting be more appropriate than paraphrasing?**

Quoting would be more appropriate in formal internal communications where exact policy language matters, such as:
- compliance reminders
- security policy updates
- audit-related communication
- mandatory training materials
- messages about rules employees must follow exactly

Quoting helps avoid changing the meaning of the policy.

**When might quoting pose a risk?**

Quoting can be risky when the original text is too formal, technical or difficult for employees to understand. It may also create confusion if the quote is taken out of context or if employees need a simpler explanation of what to do in practice.

Quoting is best when accuracy and authority are most important.
Paraphrasing is better when clarity and learning are the main goals.
