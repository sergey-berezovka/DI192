Mini project 1 : Meta-Analysis of Research Papers on Large Language Models (LLMs):
From Pipeline-Based AI Agents to Model-Native Agentic AI
1. Introduction
Large Language Models (LLMs) moved far beyond simple text generation. Early LLM applications mainly followed a reactive query-response pattern: a user asked a question, the model produced an answer, and any planning, retrieval, or tool use was usually handled by external code. The field is now entering an agentic stage, where LLMs are expected to reason over multiple steps, call tools, remember context, interact with environments, and adjust their actions toward long-term goals. The purpose of this meta-analysis is to compare recent research that explains this transition from pipeline-based agent systems to more autonomous, model-native agentic AI.
The main topic across the selected papers is the rise of agentic LLMs: systems that combine reasoning, acting, interaction, tool use, and memory.
The three papers selected are all recent surveys, which makes them especially useful for identifying broad research trends rather than evaluating one isolated model. They compare many architectures, training strategies, datasets, benchmarks, and evaluation methods used in the current agentic LLM literature.

2. Paper Summaries
2.1 Sang et al. (2025) - Beyond Pipelines: A Survey of the Paradigm Shift toward model-native agentic AI.
The research problem is how agentic AI systems are evolving from externally engineered workflows into models that internalize planning, tool use, and memory. The proposed solution is a conceptual taxonomy that contrasts pipeline-based agents with model-native agents. In the pipeline paradigm, an LLM is only one component inside a larger handcrafted workflow. In the model-native paradigm, the model itself learns to plan, call tools, and manage information through training, especially Reinforcement Learning (RL).
Reinforcement Learning is the central mechanism behind this shift. Instead of imitating static examples through supervised fine-tuning alone, Reinforcement Learning allows the model to explore trajectories and learn from outcome-based rewards. The main result is not an experimental score but a structured explanation of why modern agentic AI is moving toward the formula LLM + RL + Task. The survey discusses architectures such as reasoning models, deep research agents, and GUI agents. Datasets and benchmarks mentioned include GAIA, SWE-Bench, AndroidWorld, BrowseComp, MCP-Universe, and FutureX. Evaluation is mainly framed around task success, outcome rewards, verifiable answers, long-horizon consistency, and robustness in dynamic environments.
2.2 Plaat et al. (2025) - Agentic Large Language Models, a survey
This paper addresses the need to organize the growing field of LLM agents. Its central definition is that agentic LLMs are models that reason, act, and interact. This three-part taxonomy is broader than a purely technical view because it includes individual reasoning, external action through tools or robots, and multi-agent interaction.
The paper summarizes methods such as Chain-of-Thought, Tree-of-Thoughts, Self-Reflection, ReAct, Reflexion, retrieval augmentation, tool calling, robot action models, and multi-agent simulations. Architecturally, it treats the LLM as a core language policy that can be extended with search, memory, retrieval, tools, APIs, and interaction environments. The reviewed datasets and benchmarks include GSM8K for math reasoning, ToolBench and APIBench for tool use, WebArena and AgentBoard for web and tool-based agents, and social/game benchmarks for multi-agent behavior. Evaluation metrics vary by domain: answer accuracy, benchmark task success, tool-call correctness, code execution success, retrieval quality, and social interaction outcomes. The main result is a research agenda emphasizing that reasoning, acting, and interacting reinforce each other: retrieval supports tool use, reflection improves collaboration, and interaction can generate new training data.
2.3 Yao et al. (2025) - A Survey on Agentic Multimodal Large Language Models
This paper narrows the agentic AI discussion to multimodal large language models (MLLMs), which combine language with visual and sometimes embodied information. The research problem is that conventional MLLM agents remain static, passive, and domain-specific, while real-world multimodal tasks require adaptive planning, proactive tool invocation, and environment interaction.
The proposed framework has three dimensions: internal intelligence, external tool invocation, and environment interaction. Internal intelligence includes reasoning, reflection, and memory. Tool invocation includes search, code execution, and visual processing. Environment interaction includes virtual interfaces, GUI agents, robots, navigation, healthcare, autonomous driving, and recommender systems. Architectures discussed include dense MLLMs and mixture-of-experts MLLMs, with a trend toward MoE models because they can support adaptive reasoning and tool use more efficiently. Training strategies include continual pretraining, supervised fine-tuning on agentic trajectories, and RL methods such as PPO and GRPO. Datasets and benchmarks reviewed include MMMU-Pro, OlympiadBench, MMLongBench-Doc, MMSearch-Plus, WebMMMU, ScreenSpot, OSWorld, VLABench, and vision-language navigation benchmarks. Evaluation is divided into process evaluation, which checks reasoning steps and tool use, and outcome evaluation, which measures final task success.

3. Comparative Analysis
The key change in LLM research is not only bigger models but a new role for models. Traditional LLMs generate responses and agentic LLMs select actions. Each paper emphasizes a different level of the shift. Sang et al. focus on the system-design paradigm: external pipelines versus model-native policies. Plaat et al. focus on the functional taxonomy: reasoning, acting, and interacting. Yao et al. focus on multimodal agency: reasoning with images, tools, and environments. Together, they show that agentic AI is becoming a unified research direction rather than a collection of separate tricks:

Sang et al. are the most explicitly concerned with replacing brittle handcrafted workflows. Their main argument is that agentic abilities should be learned rather than externally scripted. Plaat et al. are more descriptive and interdisciplinary: they connect LLM agents to older AI traditions such as robotics, reinforcement learning, and multi-agent systems. Yao et al. extend the same debate into multimodal settings, where the agent must not only read text but also perceive images, browse visual interfaces, and act in virtual or physical environments.
In terms of architecture, the papers show a movement from modular systems toward integrated policies. Pipeline systems are easier to inspect because each component has a clear role: planner, retriever, tool caller, memory store, or executor. However, this modularity can make the system brittle when the task changes. Model-native systems are more adaptive because the model learns when to reason, retrieve, or act. The cost is lower interpretability and higher training complexity.
Training strategies also reveal a shared trend. Prompt engineering and supervised fine-tuning remain useful, but the papers agree that they are not enough for long-horizon agentic behavior. RL becomes important because it lets models learn from success or failure in a task environment. This is especially valuable when it is difficult to write the correct step-by-step solution in advance. Still, RL introduces new problems, including reward hacking, unstable optimization, expensive rollouts, and the difficulty of defining good rewards for open-ended tasks such as research reports.
Evaluation is another common challenge. Traditional NLP metrics such as BLEU or ROUGE are insufficient for agentic systems because the goal is not only to produce fluent text. Agentic evaluation must measure process quality, tool-call accuracy, final task completion, safety, factuality, and robustness across changing environments. The multimodal paper makes this especially clear by separating process evaluation from outcome evaluation.

4. Insights and Reflection
Several patterns emerge across the papers. First, the field is moving from prompt-level intelligence to behavior-level intelligence. A strong prompt can make a model appear agentic for a narrow task, but a real agent must decide what to do when the next step is uncertain. Second, memory is becoming a core research problem rather than a convenience feature. Without memory, agents cannot maintain long-horizon consistency or learn from previous actions. Third, tool use is becoming part of reasoning itself: future systems may not treat search, code, or visual processing as external add-ons, but as actions selected by the model policy.
The most promising direction is reinforcement-learned agentic behavior. RL is attractive because it allows models to discover strategies through interaction rather than only imitate human-written traces. This is particularly important in domains where humans cannot easily annotate every correct intermediate step, such as open-web research, software engineering, GUI control, or robotics. Multimodal agentic systems also appear highly promising because real-world tasks usually require more than text: users need agents that can inspect screens, understand documents, interpret images, and act in changing environments.
At the same time, all three papers imply serious limitations. Agentic LLMs can hallucinate, misuse tools, take unsafe actions, or optimize the wrong reward. Their long reasoning traces are expensive and difficult to verify. Reproducibility is also difficult because many frontier systems are closed-source, and agent performance can depend heavily on hidden prompts, tool access, environment design, and evaluation settings. Future research should therefore focus on safer reward design, transparent evaluation, reproducible benchmarks, better long-term memory, stronger grounding, and clearer boundaries between model autonomy and human oversight.

5. Conclusion
This meta-analysis shows that recent LLM research is shifting from static language modeling toward agentic systems capable of reasoning, acting, interacting, using tools, and managing memory. Sang et al. explain the paradigm shift from pipeline-based systems to model-native agents. Plaat et al. provide the broadest taxonomy of agentic LLMs as systems that reason, act, and interact. Yao et al. extend the discussion to multimodal settings, where agents must coordinate internal intelligence, tool invocation, and environment interaction. Across the three papers, the strongest trend is the growing role of reinforcement learning as a way to internalize agentic capabilities. The field is evolving toward models that do not merely respond to instructions but learn to pursue goals through experience. This evolution also makes safety, evaluation, interpretability, and reproducibility more important than ever.
