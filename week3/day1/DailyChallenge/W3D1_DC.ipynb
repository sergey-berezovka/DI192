**Company Economic Data Analysis:**
Analyis of company sales data to understand how discounts
affect revenue and sales performance. The goal is to demonstrate how data analysis can help businesses
make decision-making.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales, kUSD": [20000, 22000, 25000, 27000, 30000, 32000],
    "Discount, %": [5, 5, 10, 10, 15, 15],
    "Quantity of Customers": [180, 200, 220, 240, 260, 280]
}

df = pd.DataFrame(data)

# calculate revenue per customer
df["Revenue_per_customer"] = df["Sales, kUSD"] / df["Quantity of Customers"]

df. head(6)
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Month"], df["Sales, kUSD"], marker='o')

plt.title("Monthly Sales Growth")
plt.xlabel("Month")
plt.ylabel("Sales (kUSD)")
plt.grid(True)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales, kUSD": [20000, 22000, 25000, 27000, 30000, 32000],
    "Discount, %": [5, 5, 10, 10, 15, 15],
    "Quantity of Customers": [180, 200, 220, 240, 260, 280]
}

df = pd.DataFrame(data)

plt.figure(figsize=(8,5))
plt.scatter(df["Discount, %"], df["Sales, kUSD"])

plt.title("Impact of Discounts on Sales")
plt.xlabel("Discount (%)")
plt.ylabel("Sales (kUSD)")
plt.grid(True)

# подписи точек
for i in range(len(df)):
    plt.text(df["Discount, %"][i], df["Sales, kUSD"][i], df["Month"][i])

plt.show()
```

**Analysis**: The data shows that sales increase as discounts increase.
However, higher discounts also reduce profit margins.

This suggests that the company must find an optimal discount level
that increases sales while maintaining profitability.


**Conclusion**: This example demonstrates how data analysis can help companies t0 understand the relationship between discounts, customer behavior and sales performance. By analyzing business data, companies can make better strategic decisions regarding pricing, marketing and customer acquisition.
