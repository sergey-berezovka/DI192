Exercise 1: Introduction to Data Analysis (Easy):
Data analysis - process of collecting, organizing, cleaning, and interpreting data in order to extract useful information and support decision-making process.
During performing Data analysis we can identify patterns, trends, relationships, or insights.
Data analysis includes computational tools to clean and transform raw data, statistical methods and visualization techniques.
Data analysis has become essential in today’s world because organizations and individuals generate massive amounts of data every day.
By analyzing data, businesses and institutions can make informed management decisions, predict future outcomes, improve efficiency and reduce risks.
Application areas:

a) Business and Finance:
In business, data analysis helps companies understand customer preferences, optimize operation and increase profitability. Financial analysts use data to evaluate investments, forecast revenue
and assess risks. real-time analytics. Instead of analyzing data after it has been collected, companies now analyze data instantly as it is generated. This allows organizations to respond quickly
to changes.

b) Healthcare and Medicine:
In healthcare, data analysis is used to study patient data, track disease trends, and improve treatment
outcomes. Hospitals analyze medical records to identify effective treatments.The scientists are developing
vaccines and other drugs faster. During the COVID-19 pandemic, data analysis was used worldwide to track
infection rates and predict the spread of the virus.

c) Marketing and Advertising:
Data analysis helps companies understand consumer behavior and preferences. By analyzing customer data such as browsing history, purchases, and engagement with advertisements, businesses can create targeted
marketing campaigns.

Conclusion:
Including modern trends and real-world examples shows that data analysis is not only a theoretical concept but also a powerful tool used in many industries today.
As technology continues to advance, the role of data analysis will become even more important for decision-making and innovation.

1. Trend Analysis: "Year" and "Avg hrs per day sleeping"
These variables allow us to analyze how the average sleep duration of Americans has changed over time.
By plotting sleep duration against year, we can identify trends or long-term patterns in sleeping habits.

2. Group Comparison by Age: "Age Group" and "Avg hrs per day sleeping"
This comparison allows us to analyze whether different age groups sleep different amounts of time. It helps identify patterns such as whether younger or older individuals tend to sleep more.

3. Gender Comparison: "Sex and "Avg hrs per day sleeping".
These variables allow comparison of average sleep duration between males and females. This type of analysis can reveal gender-based differences in sleep patterns.

4. Weekday and Weekend comparison: "Type of Days" and "Avg hrs per day sleeping"
This analysis can show whether people sleep more on weekends compared to weekdays. It helps understand behavioral differences in sleep patterns depending on the day type.

Exercise 9 : Export a dataframe to excel format and JSON format:

```python
import pandas as pd
from google.colab import files

# Dataset:
data = {
    "Customer": ["Anna", "Sergei", "Nina", "Petr"],
    "Product": ["Laptop", "Tablet", "Smartphone", "Laptop"],
    "Price": [1200, 900, 500, 1150],
    "Quantity": [1, 2, 1, 1],
    "Purchase_Date": ["2026-03-01", "2026-03-01", "2026-03-02", "2026-03-03"]
}

df = pd.DataFrame(data)

# Files:
df.to_excel("sales_data.xlsx")
df.to_json("sales_data.json")

# Download:
files.download("sales_data.xlsx")
files.download("sales_data.json")
```
