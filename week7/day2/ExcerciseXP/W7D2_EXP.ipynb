## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- Core architecture: Bidirectional encoder-only Transformer; trained with Masked Language Modeling (MLM) (and originally Next Sentence Prediction, NSP)
- Application: Text classification (e.g., sentiment analysis) / Named Entity Recognition (NER) / Extractive Question Answering
- Why: BERT reads the entire sentence in both directions, so it captures full context. This makes it especially strong for understanding tasks where meaning depends on surrounding words
### GPT
- Core architecture: Decoder-only Transformer; autoregressive causal language model (predicts next token)
- Application: Chatbots, text generation, code generation
- Why: GPT generates text sequentially, making it ideal for tasks that require fluent, coherent output
### T5
- Core architecture: Encoder-decoder Transformer; unified text-to-text framework (everything is input => output text)
- Application: Summarization, translation, question answering
- Why: T5 treats all tasks as text generation, which makes it extremely flexible. The encoder understands input, and the decoder generates structured outputs, making it perfect for transformation tasks.


## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- Benefits (explain each in your words):
  1) Improved generalization: pretrained models learn broad language patterns from large datasets, so they perform well even on new, unseen tasks or data

  2) Less labeled data: because the model already understands language structure, it needs far fewer labeled examples to achieve good performance on a specific task

  3) Faster fine-tuning: instead of training from scratch, we only adjust a pretrained model slightly, which reduces training time and computational cost

  4) Transfer learning: knowledge learned from one task or domain can be reused for another task, improving efficiency and performance

  5) Robustness: exposure to diverse data during pretraining helps models handle noise, variation in language, and different contexts more reliably


- Ethical concerns:

  Bias: models learns and reproduces stereotypes present in training data

  Misinformation: enerate incorrect or misleading content confidently

  Misuse: used for spam, deepfakes, or harmful automation

  Privacy: risk of memorizing or leaking sensitive information from training data


- Mitigations:

  Data curation: carefully filter and balance datasets to reduce bias and harmful content

  Differential privacy: add noise or constraints during training to protect sensitive data

  Safety filters: detect and block harmful or unsafe outputs

  RLHF: align model behavior with human values
  
  Audits: regular evaluation for bias, safety, and performance issues



## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- TODO: Describe the Q/K/V flow, softmax weighting, and value mixing.

Each token is converted into three vectors: Query (Q), Key (K), and Value (V).

The query of one word is compared with the keys of all words using dot product => gives similarity scores.
These scores are scaled and passed through softmax to become attention weights (sum to 1).
Each word then becomes a weighted sum of all value vectors, meaning it “collects” information from other relevant words.


- TODO: Explain why multiple heads help (subspace projections, diverse relations).

Multi-head attention projects Q/K/V into different subspaces. Each head can focus on different types of relationships (e.g., syntax, meaning, position), allowing the model to capture richer patterns in parallel.

### Pre-training Objectives
- MLM vs CLM: TODO (compare objectives and typical use)

  MLM (Masked Language Modeling): Random words are masked, and the model predicts them using both left and right context (bidirectional). Used in encoder models like BERT

  CLM (Causal Language Modeling): Predicts the next token based only on previous tokens (left-to-right). Used in decoder models like GPT

- When to prefer MLM vs CLM:

MLM: Best for understanding tasks (classification, NER, QA) because it sees full context

CLM: Best for generation tasks (chat, writing, code) because it models natural text continuation

- NSP: TODO (why early BERT used it and why many modern models avoid it)

Originally used in BERT to learn relationships between sentences (is sentence B a continuation of sentence A?). Modern models often avoid NSP because it did not significantly improve performance; better alternatives (like larger data or different objectives) work more effectively


### Transformer Model Selection
- Sentiment on reviews: Encoder-only (BERT). Needs deep understanding of the full sentence context, not generation

- Conversational chatbot (creative responses): Decoder-only (GPT). Generates fluent, coherent text step-by-step

- Technical document translation (EN→ES): Encoder–Decoder (T5). Encoder understands source text; decoder generates structured output in another language


### Positional Encoding
- Purpose: transformers process tokens in parallel and have no built-in notion of order. Positional encoding injects information about word order (absolute or relative positions), which is essential for understanding meaning.

- Example failure without positions: “Dog bites man” vs “Man bites dog”
Without positional information, the model would treat these as identical sets of words, losing the meaning difference.

### (Optional) Quick Generation Demo Across Temperatures
Note: Requires `transformers` and model download; skip if offline.


```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline
import torch

model_name = "gpt2"  # TODO: choose a causal LM
prompt = "Artificial intelligence will transform finance by"  # TODO: write your own prompt

pipe = pipeline(
    "text-generation",
    model=AutoModelForCausalLM.from_pretrained(model_name),
    tokenizer=AutoTokenizer.from_pretrained(model_name),
    device=0 if torch.cuda.is_available() else -1,
)

for T in [0.2, 1.0, 1.5]:
    out = pipe(prompt, max_new_tokens=40, temperature=T, do_sample=True)
    print("--- Temperature:", T)
    print(out[0]["generated_text"])  # Inspect how style changes with T
```
