**Daily Challenge: Building Your First Neural Network on the MNIST Dataset:**

```python
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.utils import to_categorical

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, classification_report

print("TensorFlow version:", tf.__version__)
```

```python
# Load data
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
# Normalize
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
# One-hot encode
y_train_oh = to_categorical(y_train, 10)
y_test_oh = to_categorical(y_test, 10)
```

```python
# Model
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(32, activation="relu"),
    Dense(16, activation="relu"),
    Dense(10, activation="softmax")
])
```

```python
# Compile
model.compile(
    optimizer="adam",
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
# Train
history = model.fit(
    x_train,
    y_train_oh,
    epochs=5,
    batch_size=32,
    validation_split=0.1
)
```

```python
# Evaluate
test_loss, test_acc = model.evaluate(x_test, y_test_oh)

print("Test loss:", test_loss)
print("Test accuracy:", test_acc)
```

I implemented a simple fully connected neural network with two small hidden layers (32 and 16 neurons) using ReLU activation. The output layer uses softmax to produce probabilities over the 10 digit classes. Test loss is 0.13 and Test accuracy is 0.96.
