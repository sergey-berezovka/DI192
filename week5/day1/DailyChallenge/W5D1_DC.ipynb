**Daily Challenge: Logistic Regression for Admission Prediction**

**Data Exploration: Load and visualize the data:**

```python
import pandas as pd

url = "https://raw.githubusercontent.com/kaleko/CourseraML/master/ex2/data/ex2data1.txt"

df = pd.read_csv(url, header=None)
df.columns = ["Exam1", "Exam2", "Admitted"]

df.head()
```

```python
df.describe()
df.info()
```

**Create a scatter plot to visualize the data points for students who were admitted versus those who were not based on their exam scores:**

```python
import matplotlib.pyplot as plt

admitted = df[df["Admitted"] == 1]
not_admitted = df[df["Admitted"] == 0]

plt.scatter(admitted["Exam1"], admitted["Exam2"], label="Admitted")
plt.scatter(not_admitted["Exam1"], not_admitted["Exam2"], label="Not admitted")

plt.xlabel("Exam 1")
plt.ylabel("Exam 2")
plt.legend()
plt.show()
```

The scatter plot shows two groups of students: admitted and not admitted. Students with higher exam scores tend to be admitted, while students with lower scores are mostly not admitted.

**Applying Logistic Regression with scikit-learn, making predictions, model evaluation:**

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X = df[["Exam1", "Exam2"]]
y = df["Admitted"]

model = LogisticRegression()

model.fit(X, y)

y_pred = model.predict(X)

accuracy = accuracy_score(y, y_pred)
print("Accuracy:", accuracy)
```

The model achieved an accuracy of about 89%, meaning it correctly classifies most students.
The results indicate that exam scores are strong predictors of admission. Logistic regression successfully separates admitted and non-admitted students using a linear decision boundary based on exam scores.
